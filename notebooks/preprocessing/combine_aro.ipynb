The purpose of this notebook is to combine the ARO attribution and relation datasets into a single dataset.


```python
import os
import pandas as pd

vga_dir = '../../data/processed/aro/visual_genome_attribution'
vgr_dir = '../../data/processed/aro/visual_genome_relation'

output_dir = '../../data/processed/aro/combined'
os.makedirs(output_dir, exist_ok=True)

train_df_vga = pd.read_json(os.path.join(vga_dir, 'train.json'))
val_df_vga = pd.read_json(os.path.join(vga_dir, 'val.json'))
test_df_vga = pd.read_json(os.path.join(vga_dir, 'test.json'))

train_df_vgr = pd.read_json(os.path.join(vgr_dir, 'train.json'))
val_df_vgr = pd.read_json(os.path.join(vgr_dir, 'val.json'))
test_df_vgr = pd.read_json(os.path.join(vgr_dir, 'test.json'))

train_df = pd.concat([train_df_vga, train_df_vgr])
val_df = pd.concat([val_df_vga, val_df_vgr])
test_df = pd.concat([test_df_vga, test_df_vgr])

train_df.to_json(os.path.join(output_dir, 'train.json'), orient='records')
val_df.to_json(os.path.join(output_dir, 'val.json'), orient='records')
test_df.to_json(os.path.join(output_dir, 'test.json'), orient='records')
```
